```python
import pandas as pd
import numpy as np
from scipy.linalg import svd
```

```python
df = pd.read_csv('data/reviews.csv')
df.drop(['Unnamed: 0'], axis=1, inplace=True)
df.columns = [c.lower().replace(' ', '_') for c in df.columns]
df.info()
```

```python
df = df.reset_index(names=['customer_id']).copy()

df['is_positive'] = (df['rating'] > 3).astype(int)

df = df[['customer_id', 'clothing_id', 'is_positive']].copy()
```

```python
df
```

```python
df_onehot = df.pivot_table(index='customer_id', columns='clothing_id', 
                           values='is_positive', aggfunc='max', fill_value=0)

# Perform Singular Value Decomposition (SVD)
U, sigma, VT = svd(df_onehot.values)

# Set the number of latent factors
k = 2
sigma_k = np.diag(sigma[:k])
U_k = U[:, :k]
VT_k = VT[:k, :]

# Predict user-item ratings using the factorized matrices
predicted_ratings = np.dot(np.dot(U_k, sigma_k), VT_k)

print("Predicted user-item ratings:")
print(predicted_ratings)
```
